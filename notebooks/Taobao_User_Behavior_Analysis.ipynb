## Project Introduction

### Business Goal

The project examines how users interact with products on Taobao and identifies opportunities to improve conversion, retention, and customer engagement. The analysis connects high-level activity patterns with chronological funnels, purchase journeys, and user segments.

### Dataset Description

The processed dataset contains event-level user interactions. Each row records a timestamp, user, item, item category, and behaviour type. The four behaviours are View, Favorite, Cart, and Purchase.

### Analysis Objectives

- Summarize the scale and composition of user activity.
- Measure chronological View-to-Cart-to-Purchase conversion.
- Identify the main user-item paths leading to purchases.
- Segment users by activity and purchase frequency.
- Translate the results into practical business recommendations.

## User Behaviour Analysis

| Behaviour | Number of actions | Share of all actions |
| --- | ---: | ---: |
| View | 11,550,581 | 94.24% |
| Favorite | 242,556 | 1.98% |
| Cart | 343,564 | 2.80% |
| Purchase | 120,205 | 0.98% |

![User Behaviour Distribution](../docs/behavior_distribution.png)

### Key Observations

- Viewing accounts for 94.24% of recorded actions, so the platform activity is strongly concentrated at the discovery stage.
- Cart actions exceed favorite actions, indicating that cart usage is the more common recorded signal of purchase intent.
- Purchases represent 0.98% of actions. This is an action-level share, not a user conversion rate, so it should be interpreted alongside the chronological funnel.
- The large gap between views and downstream actions highlights the importance of product relevance, product-page quality, and low-friction purchase journeys.

## Conversion Funnel Analysis

The funnel is chronological: a valid cart action must occur after a view, and a valid purchase must occur after that cart action.

| Funnel metric | Result |
| --- | ---: |
| View users | 10,000 |
| Cart users after view | 8,538 |
| Purchase users after cart | 7,517 |
| View-to-cart conversion | 85.38% |
| Cart-to-purchase conversion | 88.04% |
| Overall conversion | 75.17% |

![Chronological User Funnel](../docs/user_funnel.png)

### Interpretation

- The funnel retains most users at both measured stages.
- The View-to-Cart stage loses 1,462 users, while the Cart-to-Purchase stage loses 1,021 users.
- The largest observed funnel opportunity is therefore before the cart stage, where product presentation and recommendation relevance can influence intent.
- The 75.17% overall rate reflects the project's chronological definition and should not be compared directly with a non-chronological behaviour ratio.

## Purchase Behaviour Path Analysis

Purchase paths are classified chronologically within each user-item pair. This preserves the sequence of behaviours leading to each recorded purchase.

| Purchase path | Purchases | Percentage |
| --- | ---: | ---: |
| View -> Purchase | 52,784 | 43.91% |
| View -> Cart -> Purchase | 40,335 | 33.56% |
| View -> Favorite -> Purchase | 4,781 | 3.98% |
| View -> Favorite -> Cart -> Purchase | 3,097 | 2.58% |
| Other | 19,208 | 15.98% |

![Purchase Behaviour Paths](../docs/purchase_behavior_paths.png)

### Key Observations

- Direct View-to-Purchase is the largest classified path at 43.91%.
- View-to-Cart-to-Purchase contributes a further 33.56%; together, the two leading paths account for 77.47% of purchases.
- Favorite-based paths account for a smaller share of purchases than cart-based journeys.
- Other paths represent 15.98%, making them a meaningful target for more detailed path decomposition.

## User Segmentation

Users are classified using percentile-based activity and purchase-frequency thresholds. The four segments are mutually exclusive, and every user belongs to exactly one segment.

| User segment | Users | Percentage |
| --- | ---: | ---: |
| High-Frequency Buyer | 1,815 | 18.15% |
| Regular Buyer | 6,117 | 61.17% |
| High Activity Low Purchase | 954 | 9.54% |
| No Purchase | 1,114 | 11.14% |

![User Segmentation](../docs/user_segmentation.png)

### Segment Interpretation

- **High-Frequency Buyer:** The strongest repeat-purchase group and the main audience for loyalty and retention strategies.
- **Regular Buyer:** The largest segment and the main opportunity for increasing purchase frequency through relevant recommendations.
- **High Activity Low Purchase:** Users who are engaged but convert less frequently, making them suitable for targeted reminders and friction-reduction experiments.
- **No Purchase:** Users who have not purchased and may require first-purchase onboarding, trust-building, or stronger product relevance.

## Business Insights

### User Behaviour Patterns

- User activity is dominated by product views, while cart and purchase actions form a much smaller share of total events.
- Direct purchase after viewing is common, but cart-based journeys are also central to conversion.
- Most users are buyers, with Regular Buyers forming the largest segment.

### Conversion Opportunities

- Prioritize the View-to-Cart stage because it contains the largest observed user loss in the chronological funnel.
- Investigate high-view items and categories that fail to progress to cart or purchase.
- Decompose the Other purchase-path category to identify additional journeys and data-quality issues.

### Possible Marketing Strategies

- **High-Frequency Buyers:** Use loyalty benefits, personalized recommendations, and retention monitoring.
- **Regular Buyers:** Encourage repeat purchases through relevant cross-selling and category-based recommendations.
- **High Activity Low Purchase:** Test browse reminders, cart reminders, and carefully targeted incentives.
- **No Purchase Users:** Improve onboarding, trust signals, and first-purchase messaging.

> Recommendations should be validated through controlled experiments. The current analysis is descriptive and does not establish causality.

## Future Improvements

- Validate duplicate records and quantify their effect before changing the cleaning process.
- Break the Other purchase-path category into more specific chronological sequences.
- Compare behaviour, funnels, and segments across categories and time periods.
- Add price or transaction-value data to support revenue, average order value, and monetary customer-value analysis.
- Evaluate how sensitive user segments are to different percentile thresholds.
- Extend automated tests to additional funnel edge cases and reporting outputs.
- Automate report generation after analytical definitions and data-quality rules are finalized.

## Notes and Limitations

- The processed CSV is approximately 492 MB and is intentionally not embedded in this notebook.
- Duplicate rows were detected but were not automatically deleted because their business meaning requires validation.
- The dataset has no price field, so revenue and monetary customer value cannot be calculated.
- Purchase paths represent recorded events within the available observation window.
- User segments are relative to this dataset because their thresholds are percentile-based.
